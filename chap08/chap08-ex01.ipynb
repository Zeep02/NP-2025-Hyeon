**연습문제 1. 사용자로부터 두 개의 정수를 입력받아 나눗셈을 수행하되, 0으로 나누는 경우와 정수가 아닌 값을 입력하는 경우를 예외처리하라.**


```python
try:
    a = int(input("첫 번째 정수: "))
    b = int(input("두 번째 정수: "))
    result = a / b
    print(f"{a} / {b} = {result:.2f}")
except ValueError:
    print("정수를 입력해주세요.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
finally:
    print("프로그램 종료")
```

**연습문제 2. 파일 이름을 입력받아 해당 파일의 내용을 출력하는 프로그램을 작성하라. 파일이 존재하지 않는 경우를 예외처리하라.**


```python
filename = input("파일 이름을 입력하세요: ")
try:
    with open(filename, "r", encoding="utf-8") as f:
        content = f.read()
        print("\\n파일 내용:")
        print(content)
except FileNotFoundError:
    print(f"'{filename}' 파일을 찾을 수 없습니다.")
```

**연습문제 3. 학생의 이름과 점수를 입력받아 파일에 저장하고, 저장된 파일을 읽어서 출력하는 프로그램을 작성하라.**


```python
# 파일에 쓰기
with open("students.txt", "w", encoding="utf-8") as f:
    for i in range(3):
        name = input(f"{i+1}번째 학생 이름: ")
        score = input(f"{i+1}번째 학생 점수: ")
        f.write(f"{name},{score}\\n")

print("\\n저장된 학생 정보:")
# 파일에서 읽기
with open("students.txt", "r", encoding="utf-8") as f:
    for line in f:
        name, score = line.strip().split(",")
        print(f"이름: {name}, 점수: {score}")
```
