### 예외처리

**예외(exception)란?**
- 프로그램 실행 도중에 오류가 발생하면 생성되는 특별한 객체
- 예외가 발생하면 파이썬은 프로그램 실행을 중단한다
- 예외처리란 예외가 발생했을 때 프로그램 실행을 중단하지 않고 처리하는 방법

**예외의 종류**
- `ZeroDivisionError`: 0으로 나눌 때
- `IndexError`: 인덱스가 범위를 벗어날 때
- `FileNotFoundError`: 존재하지 않는 파일을 열려고 할 때
- `ValueError`: 원하는 값을 입력받지 못할 때
- `NameError`: 정의되지 않은 변수를 사용할 때
- `TypeError`: 데이터형이 맞지 않는 연산을 할 때


### try-except 문

**기본 형식**
```python
try:
    # 실행할 코드
except 예외종류:
    # 예외 처리 코드
else:
    # 예외가 발생하지 않으면 실행
finally:
    # 항상 실행
```


````python
# try-except 예제
try:
    b = 0
    a = 1 / b
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다")
else:
    print(a)
````

````python
# 예외를 알 수 없는 경우
try:
    a = int(input("Type a Number: "))
except Exception as e:
    print("예외가 발생했습니다", e)
else:
    print(a)
````

````python
# try-except-finally 예제
try:
    x = 10
    y = 0
    result = x / y
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다")
finally:
    print("예외처리가 끝났습니다")
````

### 파일 처리

**파일 입출력**
- **표준 입력**: 키보드
- **표준 출력**: 콘솔 화면
- **텍스트 파일**: 사람이 볼 수 있는 문자 데이터
- **이진 파일**: 이진수 데이터

**파일 처리 과정**
1. 열기: 파일 이름과 처리 목적을 지정해 파일 열기
2. 사용: 열린 파일에 대한 읽기 또는 쓰기 진행
3. 닫기: 파일 사용이 끝나면 닫기


````python
# 텍스트 파일 쓰기
ofile = open("./test.txt", "w", encoding="utf-8")
ofile.write("Hello, World!\\n")
ofile.write("Python Programming\\n")
ofile.close()
````

````python
# 텍스트 파일 읽기
try:
    rfile = open("./test.txt", "r", encoding="utf-8")
    line = rfile.readline()
    while line:
        print(line.strip())
        line = rfile.readline()
    rfile.close()
except FileNotFoundError as e:
    print("File not found:", e)
````

### with 문

**with~as 구문**
- 파일을 자동으로 닫아주는 구문
- 예외가 발생해도 자동으로 파일을 닫음


````python
# with 문 사용 예제
with open("./test.txt", "w", encoding="utf-8") as wfile:
    wfile.write("Line 1\\n")
    wfile.write("Line 2\\n")

with open("./test.txt", "r", encoding="utf-8") as rfile:
    for line in rfile:
        print(line.strip())
````

### F-String

**F-String (formatted string)**
- 변수나 표현식으로 내용을 변경할 수 있는 포맷 문자열
- Python 3.6+에서 사용 가능


````python
# F-String 기본 사용
name = "Python"
age = 30
print(f"My name is {name} and I am {age} years old")

# 수식 계산
a = 10
b = 3
print(f"{a} + {b} = {a+b}")
print(f"{a} * {b} = {a*b}")
````

````python
# F-String 정렬과 포맷
pi = 3.14159
print(f"PI = {pi:.2f}")  # 소수점 2자리

text = "Python"
print(f"'{text:<10}'")  # 왼쪽 정렬
print(f"'{text:>10}'")  # 오른쪽 정렬
print(f"'{text:^10}'")  # 가운데 정렬

num = 1234567
print(f"{num:,}")  # 쉼표로 구분
````
