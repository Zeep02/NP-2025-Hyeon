### 프로그램 제어문


**프로그램 제어문**
- 프로그램은 작성 순서대로 실행
- 제어문: 프로그램의 실행 순서를 바꾸는 문장

### if 제어문


**if 문**
- 조건을 만족할 때 정해진 문장 블록을 실행
- 조건이 참(True)이면 블록을 실행하고 아니면 블록을 건너뛴다

```python
if 조건:
    블록
```


````python
# if 문 예제
score = int(input("정수 입력: "))
if score >= 90:  # if문 끝에 :을 붙인다
    print("성적: A")  # 들여쓰기
    print("장학금 수여")
````

````python
# 복합 조건문
age_0 = 22
age_1 = 18
age = int(input('your age? '))
if age < age_0 and age > age_1:  # 복합 조건문
    print("당신의 나이는 18~22세 사이입니다")
````

**if-else 문**
- 조건이 참(True)이면 Block1을 실행하고 else:는 무시
- 참(True)이 아니면 else: 다음 Block2를 실행
- 한 줄로 표현하기도 가능

```python
if 조건:
    블록1
else:
    블록2
```


````python
# if-else 문 예제
score = 85
if score >= 90:
    print("pass")
else:  # else도 끝에 :을 붙인다
    print("fail")  # 들여쓰기

# 한 줄로 표현
res = 'pass' if score >= 90 else 'fail'
print(res)
````

**평년과 윤년 구분하는 프로그램**
- 년도가 4로 나뉘어지고 100으로 나뉘어지지 않거나
- 400으로 나뉘어지면 윤년이다


````python
# 윤년 판별 프로그램
year = int(input("Type a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400) == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
````

**중첩 if 문**
- if 문 속에 또 다른 if 문


````python
# 중첩 if 문 - 수의 판별: 양수, 음수, 0
n = int(input("type int number: "))
if n > 0:
    print("positive")
else:
    if n < 0:
        print("negative")
    else:
        print("0")
````

````python
# 중첩 if 문 - 다른 방법
n = int(input("type int number: "))
if n >= 0:
    if n == 0:
        print("0")
    else:
        print("positive")
else:
    print("negative")
````

**if-elif-else 문**
- if 조건1이 참이면 블록1, if 조건1이 거짓이고 elif 조건2이 참이면 블록2, 조건이 모두 거짓이면 블록3이 실행된다

```python
if 조건1:
    블록1
elif 조건2:
    블록2
else:
    블록3
```


````python
# if-elif-else 문 - 수의 판별
n = int(input("type int number: "))
if n > 0:
    print("positive")
elif n < 0:
    print("negative")
else:
    print("0")
````

**다중 elif 문**
- 여러 개의 elif를 사용한 문장
- if 또는 elif 조건 중 하나만 충족되면 나머지는 무시


````python
# 입장료 계산 프로그램
age = 12
if age < 4:
    price = 0
elif age < 18:
    price = 5
elif age < 65:
    price = 10
else:
    price = 5
print("Your admission cost is $" + str(price) + ".")
````

### while 반복문


**while 문**
- 조건이 충족되면 동일한 일을 반복 수행
- 조건이 만족되면 블록을 실행하고 다시 조건을 조사
- 조건이 만족되지 않으면 다음 문장 실행

```python
while 조건:
    블록
```


````python
# 1~100까지의 합을 구하는 프로그램
i = 1  # 초기화
sum_val = 0
while i <= 100:  # 조건 판단. 끝에 :를 붙인다
    sum_val = sum_val + i  # 블록
    i = i + 1  # 조건 변경
print("sum =", sum_val)  # 블록이 끝나면 다시 들여쓰기를 원래 위치로
````

**10진수를 2진수로 변환하는 프로그램**


````python
# 10진수를 2진수로 변환
n = int(input('Number: '))
result = ''  # 변환 결과 저장
# 10진수를 2로 나누어 몫이 0이 될 때까지 반복해서 나머지를 구한다
while n != 0:
    m = n % 2  # 나머지
    result = str(m) + result  # 숫자를 문자열로 변환
    n = n // 2  # 몫
print("2진수 =", result)
````

### for 제어문


**for 문**
- [반복객체]의 각 요소의 값을 하나씩 [변수]에 저장하고 블록을 실행
- [반복객체]에 더 이상의 요소가 없을 때까지 반복
- 튜플, 문자열과 같은 반복객체를 사용할 수도 있다

```python
for 변수 in 반복객체:
    블록
```


````python
# for 문 예제
for i in [1, 2, 3]:  # 끝에 :를 붙인다
    print("숫자", i)

for i in (1, 2, 3):
    print("숫자", i)

for i in 'hello':
    print(i)
````

**for-range 문**
- range() 함수는 반복객체(iterable)를 생성하는 함수
  - range(start, stop, step)
  - 예) list(range(1, 5, 1)) → [1, 2, 3, 4]
- 변수 값을 start부터 시작하여 step 만큼 증가시키며 stop될 때까지 실행
- step=1이면 생략 가능


````python
# for-range 문
for i in range(1, 4, 1):  # 1부터 시작하여 1씩 증가하여 4가 되면 종료
    print("Number", i)
````

````python
# for 문을 이용한 1부터 100까지의 합 구하기
sum_val = 0
for i in range(1, 101, 1):
    sum_val = sum_val + i
print("sum =", sum_val)
````

````python
# x^n을 구하는 프로그램
x = float(input('Type x: '))
n = int(input('Type n: '))
prod = 1
for i in range(1, n+1):
    prod = prod * x
print(prod)
````

**숫자 맞추기 게임 프로그램**
- 1~100 사이의 난수를 만들고 사용자가 1~100 사이의 수를 입력하여 맞추는 프로그램
- 5회까지 맞추지 못하면 종료


````python
# 숫자 맞추기 게임
from random import randint

secret_num = randint(1, 100)
num_guesses = 0  # 시도 회수
guess = 0  # 예상 숫자

while guess != secret_num and num_guesses <= 4:
    guess = eval(input('Enter your guess (1-100): '))  # 수 입력
    num_guesses = num_guesses + 1  # 시도 횟수 증가
    if guess < secret_num:  # 정답이 클 때
        print('더 큽니다.', 5-num_guesses, '회 남았습니다.\n')
    elif guess > secret_num:  # 정답이 작을 때
        print('더 작습니다.', 5-num_guesses, '회 남았습니다.\n')
    else:  # 정답
        print('맞았습니다!')

if num_guesses == 5 and guess != secret_num:  # 맞추지 못한 경우
    print('당신이 졌습니다. 정답은', secret_num, '입니다')
````

### 기타 제어문


**break 문**
- 반복문을 빠져나올 때 사용


````python
# break 문
n = 10
while n >= -10:
    if n == 0:
        break  # n=0이면 while 블록을 빠져나온다
    inv = 1.0 / n
    print(inv)
    n = n - 1
````

**continue 문**
- 블록의 나머지 부분은 생략하고 다시 while 조건 조사


````python
# continue 문
n = 10
while n >= -10:
    if n == 0:
        n = n - 1
        continue  # n=0이면 while 블록의 나머지 부분을 건너뛴다
    inv = 1.0 / n
    print("n = %d, inv(n) = %0.2f" % (n, inv))
    n = n - 1
````

**match 문 (Python 3.10+)**
- C에서의 switch ~ case ~ default 문장
- Python 3.10부터 추가됨

```python
match value:
    case pattern1:
        # 실행할 코드
    case pattern2:
        # 실행할 코드
    case _:  # default case (와일드카드)
        # 기본 실행할 코드
```


````python
# match 문 예제 (Python 3.10+)
def handle_status(status):
    match status:
        case 200:
            return "성공"
        case 404:
            return "페이지를 찾을 수 없음"
        case 500:
            return "서버 오류"
        case _:
            return "알 수 없는 상태"

print(handle_status(200))  # "성공"
print(handle_status(999))  # "알 수 없는 상태"
````
