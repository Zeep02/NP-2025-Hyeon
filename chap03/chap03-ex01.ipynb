**연습문제 1. 두 개의 정수를 입력받아 큰 수를 출력하는 프로그램을 작성하라.**


````python
a = int(input("첫 번째 정수: "))
b = int(input("두 번째 정수: "))

if a > b:
    print(f"큰 수는 {a}입니다.")
else:
    print(f"큰 수는 {b}입니다.")
````

**연습문제 2. 점수를 입력받아 90점 이상이면 A, 80점 이상이면 B, 70점 이상이면 C, 60점 이상이면 D, 그 미만이면 F를 출력하는 프로그램을 작성하라.**


````python
score = int(input("점수를 입력하세요: "))

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f"학점: {grade}")
````

**연습문제 3. 1부터 n까지의 짝수의 합을 구하는 프로그램을 작성하라.**


````python
n = int(input("n을 입력하세요: "))
sum_even = 0

for i in range(1, n+1):
    if i % 2 == 0:
        sum_even += i

print(f"1부터 {n}까지의 짝수의 합: {sum_even}")
````

**연습문제 4. 구구단 중 원하는 단을 입력받아 출력하는 프로그램을 작성하라.**


````python
dan = int(input("구구단 몇 단을 출력할까요? "))

print(f"\n{dan}단:")
for i in range(1, 10):
    print(f"{dan} x {i} = {dan * i}")
````

**연습문제 5. 별(*)을 이용하여 다음과 같은 삼각형을 출력하는 프로그램을 작성하라.**
```
*
**
***
****
*****
```


````python
n = int(input("삼각형 높이를 입력하세요: "))

for i in range(1, n+1):
    print('*' * i)
````
