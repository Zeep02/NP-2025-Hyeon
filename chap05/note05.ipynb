### 파이썬의 함수


**함수란?**
- 이름을 붙여 반복적으로 사용할 수 있는 프로그램 부분
- print(), input(), int(), str() 등

**함수를 사용하는 이유**
- 프로그램 안에서 중복된 코드를 제거한다
- 복잡한 프로그래밍 작업을 더 간단한 작업들로 분해할 수 있다
- 함수는 한번 만들어지면 다른 프로그램에서도 재사용될 수 있다

**파이썬의 함수는 일급 함수이다** (함수를 변수처럼 사용)

**함수의 종류**

### 함수의 정의와 호출

````python
# 함수의 정의와 호출 예
def welcome(name):  # 함수의 정의
    print("Hello,", name)  # return 문이 없어도 됨. None 반환

welcome('John')  # 함수의 호출. 'John'이 매개변수로 전달됨
````

### 함수의 반환

**반환값이 없는 함수**


````python
# 반환값이 없는 함수
def prtStr(str):  # str을 매개변수라 함
    print("%s" % str)
    return  # 반환값이 없이 함수가 종료

str_val = "Welcome to Python"
prtStr(str_val)  # 매개변수를 전달하여 함수 호출
````

**값을 반환하는 함수**


````python
# 값을 반환하는 함수
def squareArea(s):
    area = s * s
    return area  # area 값이 반환되고 종료

a = squareArea(5)  # 5를 인자(argument)라 함. 반환값이 a에 저장
b = squareArea(7)
print("한 변의 길이가 %d인 정사각형의 넓이는 %d" % (5, a))
print("한 변의 길이가 %d인 정사각형의 넓이는 %d" % (7, b))
````

### 함수의 예

**1~n까지의 합을 계산하는 함수**


````python
# 1~n까지의 합을 계산하는 함수
def sum_func(n):  # 매개변수 n을 전달받는다
    hab = 0
    for i in range(1, n+1):  # 1~n까지 반복
        hab = hab + i  # 합에 새로운 수를 더한다
    return hab  # 반환값 = hab

a = sum_func(50)  # 함수를 호출하여 (인자=50) 1부터 50까지의 합을 계산
b = sum_func(1000)  # 함수를 호출하여 (인자=1000) 1부터 1000까지의 합
print(a)
print(b)
````

### 함수와 변수

**지역변수(local variable)와 전역변수(global variable)**
- **지역변수**: 함수 내에서만 값이 유효
- **전역변수**: 프로그램 전체에서 값이 유효


````python
# 지역변수 예제
def func1():
    v = 10  # 함수 func1() 안의 지역변수. func1 내에서만 유효
    print("func1()의 v = %d" % v)

def func2():
    v = 20  # 함수 func2() 안의 지역변수. func2 내에서만 유효
    print("func2()의 v = %d" % v)

# 메인 코드 부분
func1()
func2()
````

**전역변수의 유효범위**


````python
# 전역변수의 유효범위
def func1():
    v = 10  # 함수 func1() 안의 지역변수. 지역변수가 우선 사용
    print("func1()의 v = %d" % v)

def func2():
    print("func2()의 v = %d" % v)  # 함수 func2() 안의 전역변수

# 전역변수
v = 20  # 전역변수이므로 프로그램 내에서 유효

# 메인 코드 부분
func1()  # 지역변수 값인 10을 출력. 지역변수가 없을 때 전역변수 사용
func2()  # 전역변수 값인 20을 출력
````

**전역변수의 선언 (global)**


````python
# 전역변수의 선언
def func1():
    v = 10  # 함수 func1() 안의 지역변수
    print("func1()의 v = %d" % v)

def func2():
    global v  # v를 전역변수로 선언
    v = 30
    print("func2()의 변경된 전역 v = %d" % v)

v = 20  # 함수 밖에서 정의되었으므로 전역변수

# 메인 코드 부분
func1()  # 10을 출력
func2()  # 30을 출력
print(v)  # 30을 출력. 전역변수 v가 변경됨
````

### 함수 매개변수의 전달

**매개변수의 기본값 지정 (기본 매개변수)**
- 일반(위치) 매개변수의 값을 전달하지 않으면 오류


````python
# 기본 매개변수
def prtMesg(message, count=1):  # count의 기본값이 1로 지정
    print(message * count)

prtMesg('default')  # message='default'. count는 기본값 사용
prtMesg('Hi! ', 5)  # message와 count 인자 전달

def greet(name, msg="Nothing new?"):
    print("Hi!", name + ', ' + msg)

greet("Abe")  # name만 전달
greet("Bob", "Good morning!")  # name과 msg 전달
````

**가변 매개변수** - 매개변수의 개수가 가변일 때
- 매개변수가 *arg로 표시되면 튜플로 처리
- 매개변수가 **keywords로 표시되면 딕셔너리로 처리


````python
# 가변 매개변수 - 튜플
def total(*numbers):
    sum_val = 0
    for n in numbers:  # 전달받은 매개변수를 튜플로 처리
        sum_val += n
    return sum_val

print(total(1, 2, 3))  # 3개의 인자
print(total(1, 2, 3, 4, 5))  # 5개의 인자
````

````python
# 가변 매개변수 - 딕셔너리
def dicPresident(**keywords):
    for i in keywords.keys():  # 딕셔너리 key를 하나씩 처리
        print("%s : %d-th president" % (i, keywords[i]))

dicPresident(Kennedy=35, Obama=44, Trump=45)  # (key=value)로 전달
````

**키워드를 이용한 매개변수 전달**
- 매개변수의 위치가 아니라 키워드를 이용하여 매개변수 전달


````python
# 키워드를 이용한 매개변수 전달
def func(a, b=2, c=3):  # a: 위치 매개변수, b, c: 키워드 매개변수
    print('a=', a, 'b=', b, 'c=', c)
    return a + b + c

print(func(4, 5))  # 인자 위치에 의해 a=4, b=5로 전달
print(func(5, c=7))  # 키워드 c를 이용한 전달. a=5, c=7, b는 기본값
````

### 람다(lambda) 함수

**람다(무명) 함수**
- 이름이 없는 한 줄짜리 함수
- `lambda parameters : expr`  (lambda 함수의 정의)
- parameters는 매개변수, expr은 결과가 return되는 수식
- `lambda x, y: x + y`  (매개변수 x, y를 전달받아 x + y를 리턴)
- 두 개 이상의 값을 리턴하려면 튜플, 리스트 사용
- `lambda x, y: (x + y, x * y)`  ((x+y, x*y) 튜플을 반환)


````python
# 람다 함수
sum_lambda = lambda x, y: x + y  # lambda는 함수다!
print("정수의 합:", sum_lambda(10, 20))  # 10+20을 반환
print("정수의 합:", sum_lambda(20, 20))  # 20+20을 반환

# 람다 함수의 호출
print((lambda x, y: x + y)(1, 2))  # 3
````

### 재귀함수

**재귀함수(recursion function)**
- 함수 내에서 자기 자신을 호출하는 함수
- 같은 동작을 반복할 때 재귀함수 사용
- 함수가 종료되는 조건을 반드시 포함해야 한다
- factorial(n!)을 구하는 재귀함수


````python
# 재귀함수 - factorial
def fact(n):
    if n == 1:  # 재귀함수 종료 조건. n=1이면 종료
        return 1
    else:
        return n * fact(n - 1)  # n! = n x (n-1)!

print(fact(5))
````

### 내장 함수

**eval**
- `eval(expr, [globals[, locals]])`
- 문자열로 표시된 expr의 계산 결과를 반환. 계산할 수 없으면 오류


````python
# eval 함수
print(eval('1+2*3'))  # 7
````

**int, bin, oct, hex**
- `int(str, base=10)`
- 문자열 str을 10진수로 변환


````python
# int 함수
print(int('123'))  # 123
print(int('1010', 2))  # '1010'을 2진수로 이해 -> 10
````

**filter**
- `filter(func, seq)`
- seq의 각 요소에 대해 func를 호출하여 결과 True인 요소만 모아서 iterable로 반환


````python
# filter 함수
result = list(filter(lambda x: len(x) > 2, ['this', 'is', 'a', 'test']))
print(result)  # ['this', 'test'] - 리스트 요소 중에서 길이가 2보다 큰 요소만 모음
````

**map**
- `map(func, seq, ...)`
- seq의 각 요소에 대해 func를 호출하여 실행하고 결과를 iterable로 반환


````python
# map 함수
def Squares(n):
    return n**2

numbers = [1, 3, 5, 9]
# Squares 함수에 numbers 요소를 하나씩 대입
result = list(map(Squares, numbers))
print(result)  # [1, 9, 25, 81]
````

**ord와 chr**
- `ord(ch)` – ordinal의 약자(순서가 있는 정수값)
- ch에 대한 ASCII 코드 반환(유니코드 문자)
- `ord('A')` → 65
- `hex(ord("한"))` → 0xd55c


````python
# ord와 chr 함수
print(ord('A'))  # 65
print(chr(65))  # 'A'
print(hex(ord("한")))  # 0xd55c
````

**round**
- `round(x, n=0)`
- 실수 x를 소수점 아래 n자리로 반올림하여 반환


````python
# round 함수
print(round(12.345, 1))  # 12.3
print(round(12.345, 2))  # 12.35
````

**zip**
- `zip(seq, *seqs)`
- seq 요소와 *seqs 요소의 튜플 쌍으로 만들어 리스트 iterable 반환


````python
# zip 함수
names = ['Peter Parker', 'Clark Kent', 'Wade Wilson', 'Bruce Wayne']
heroes = ['Spiderman', 'Superman', 'Deadpool', 'Batman']
for name, hero in zip(names, heroes):
    print(f'{name} is actually {hero}')

# 튜플 리스트로 만들기
print(list(zip(names, heroes)))
````

### 타입이 있는 함수와 변수 정의

**명시적 변수 타입, 함수의 반환 타입**
- Python 3.5 버전(2015년 9월 이후 지원하기 시작)
- 그 이전 버전의 코드에는 타입을 지정할 수 없음

**타입 힌트의 장점**
- 코드의 가독성 향상: 함수의 입력값과 반환값 타입을 명확히 하여 코드 이해가 쉬워집니다
- 버그 예방: 타입 불일치로 인한 오류를 사전에 발견할 수 있습니다
- 도구 지원: 타입 검사 도구(mypy 등)를 활용해 정적 분석이 가능해집니다
- 대규모 프로젝트에서 유지보수성 강화: 여러 개발자가 협업할 때 타입 정보를 통해 의도를 명확히 전달할 수 있습니다


````python
# 동적 타입 (타입 힌트 없음)
def add(a, b):
    return a + b

def greet(name):
    return f"Hello, {name}!"

# 명시적 타입 (타입 힌트 있음)
def add_typed(a: int, b: int) -> int:
    return a + b

def greet_typed(name: str) -> str:
    return f"Hello, {name}!"

def get_list() -> list:
    return [1, 2, 3]

print(add(10, 20))
print(add_typed(10, 20))
print(greet("Python"))
print(greet_typed("Python"))
````
