**연습문제 1. 두 개의 매개변수 n, m을 전달받아 m x n개의 * 상자를 출력하는 프로그램을 함수로 작성하라.**


```python
def print_stars(n: int, m: int) -> None:
    """m x n개의 * 상자를 출력하는 함수"""
    for i in range(n):
        print('*' * m)

# 테스트
print_stars(2, 4)
```

**연습문제 2. 하나의 숫자를 전달받아 숫자의 자리합을 구하는 함수를 작성하라.**  
예: 123 → 1+2+3 = 6


```python
def digit_sum(n: int) -> int:
    """숫자의 자리합을 구하는 함수"""
    total = 0
    n = abs(n)  # 음수인 경우 양수로 변환
    while n > 0:
        total += n % 10
        n = n // 10
    return total

# 테스트
print(digit_sum(123))  # 6
print(digit_sum(456))  # 15
```

**연습문제 3. 두 개의 문자열이 서로 다른 처음 위치를 반환하는 함수를 작성하라. 두 개의 문자열이 같으면 -1을 반환하라.**


```python
def find_first_diff(str1: str, str2: str) -> int:
    """두 문자열이 서로 다른 처음 위치를 반환하는 함수"""
    min_len = min(len(str1), len(str2))
    
    for i in range(min_len):
        if str1[i] != str2[i]:
            return i
    
    # 한 문자열이 다른 문자열의 접두사인 경우
    if len(str1) != len(str2):
        return min_len
    
    return -1  # 두 문자열이 같은 경우

# 테스트
print(find_first_diff("hello", "world"))  # 0
print(find_first_diff("python", "python"))  # -1
print(find_first_diff("abc", "abd"))  # 2
```

**연습문제 4. 숫자를 전달받아 그 수의 약수를 리스트로 반환하는 함수를 작성하라.**


```python
def get_divisors(n: int) -> list:
    """숫자의 약수를 리스트로 반환하는 함수"""
    divisors = []
    for i in range(1, n + 1):
        if n % i == 0:
            divisors.append(i)
    return divisors

# 테스트
print(get_divisors(12))  # [1, 2, 3, 4, 6, 12]
print(get_divisors(20))  # [1, 2, 4, 5, 10, 20]
```

**연습문제 5. 문자열과 하나의 문자를 전달받아 문자열에서 문자의 위치를 모두 찾아 리스트로 반환하는 함수를 작성하라.**


```python
def find_char_positions(text: str, char: str) -> list:
    """문자열에서 특정 문자의 위치를 모두 찾아 리스트로 반환하는 함수"""
    positions = []
    for i in range(len(text)):
        if text[i] == char:
            positions.append(i)
    return positions

# 테스트
print(find_char_positions("hello world", "l"))  # [2, 3, 9]
print(find_char_positions("python programming", "p"))  # [0, 7]
```

**연습문제 6. 재귀함수를 이용하여 1부터 100까지의 합을 계산하는 프로그램을 작성하라.**


```python
def recursive_sum(n: int) -> int:
    """재귀함수를 이용하여 1부터 n까지의 합을 계산하는 함수"""
    if n == 1:
        return 1
    else:
        return n + recursive_sum(n - 1)

# 테스트
print(recursive_sum(100))  # 5050
print(recursive_sum(10))   # 55
```
