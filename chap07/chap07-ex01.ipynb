**연습문제 1. random 모듈을 사용하여 주사위를 10번 던지는 프로그램을 작성하라.**


```python
import random

print("주사위 10번 던지기:")
for i in range(10):
    dice = random.randint(1, 6)
    print(f"{i+1}번째: {dice}")
```

**연습문제 2. math 모듈을 사용하여 반지름을 입력받아 원의 면적을 계산하는 프로그램을 작성하라.**


```python
import math

radius = float(input("원의 반지름을 입력하세요: "))
area = math.pi * radius ** 2
print(f"원의 면적: {area:.2f}")
```

**연습문제 3. datetime 모듈을 사용하여 현재 날짜와 시간을 출력하는 프로그램을 작성하라.**


```python
from datetime import datetime

now = datetime.now()
print(f"현재 날짜: {now:%Y년 %m월 %d일}")
print(f"현재 시간: {now:%H시 %M분 %S초}")
print(f"요일: {now:%A}")
```
