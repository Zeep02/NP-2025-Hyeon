### 모듈

**모듈이란?**
- 함수, 변수, 클래스들을 모아놓은 파일 묶음(라이브러리)

**모듈의 종류**
- 표준 모듈
- 서드파티(3rd party) 모듈
- 사용자 정의 모듈

**파이썬은 오픈소스를 지향**
- 많은 사람들이 개발하여 공개한 수많은 모듈이 존재
- 오픈 모듈은 무료로 사용 가능
- 빅데이터, 인공지능 등 4차 산업 프로그램들은 오픈 모듈을 사용


### 모듈 사용하기

**모듈 불러오기 방법**

1. `import module_name`
2. `from module_name import class or methods`
3. `from module_name import *`
4. `import module_name as alias`


````python
# 모듈 불러오기 방법 1
import socket
print(socket.gethostname())  # 모듈이름.속성 형태로 호출
````

````python
# 모듈 불러오기 방법 4 - alias 사용
import socket as so
print(so.gethostname())  # 단축 이름으로 속성 호출
````

### 사용자 정의 모듈

**사용자 정의 모듈 만들기**
- 모듈은 일반 프로그램과 같이 작성
- `if __name__ == '__main__'`: 모듈을 직접 실행할 때만 실행됨


### 유용한 모듈들

**sys 모듈**
- 파이썬 인터프리터에 관한 정보


````python
import sys
print(sys.version)  # 파이썬 버전 정보
print(sys.prefix)   # 파이썬 설치 경로
````

**random 모듈**
- 난수 생성


````python
import random
print(random.randint(1, 5))     # 1~5 사이의 임의 정수
print(random.random())          # 0~1 사이의 실수
print(random.choice('hello'))   # 'hello'에서 임의로 하나의 문자
````

**math 모듈**
- 수학적 계산


````python
import math
print(math.pi)                  # pi 상수값
print(math.sin(math.pi / 6))    # sin 함수
print(math.sqrt(16))            # 제곱근
````

**time과 calendar 모듈**


````python
import time
import calendar

print(time.time())      # 1970.1.1부터 경과한 초
print(time.asctime())   # 현재 날짜와 시간
print(calendar.month(2025, 10))
````

### 패키지 (Package)

**모듈과 패키지의 차이**

| 구분 | 모듈 | 패키지 |
|------|------|--------|
| 구성 단위 | 하나의 .py 파일 | 디렉토리 |
| 초기화 파일 | 필요 없음 | `__init__.py` 필요 |
| 용도 | 코드 재사용 | 모듈 그룹화 |


### 파이썬 가상환경

**가상환경이란?**
- 프로젝트별로 독립된 파이썬 환경을 만드는 도구
- 각 프로젝트가 서로 다른 패키지 버전 사용 가능

**만드는 방법**
1. venv (파이썬 3.3+ 내장)
2. conda (anaconda)
3. virtualenv (서드파티)


**venv 사용법**

```bash
# 가상환경 생성
python -m venv .myvenv

# Windows에서 활성화
.\\.venv\\Scripts\\activate

# 비활성화
deactivate

# 패키지 저장 및 설치
pip freeze > requirements.txt
pip install -r requirements.txt
```
