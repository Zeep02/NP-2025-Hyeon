### 클래스 개요
- **클래스(class)**: 객체를 생성하기 위한 템플릿 또는 설계도
- **객체(object)**: 클래스로부터 생성된 실체, 인스턴스(instance)라고도 함
- **목적**: 코드의 재사용성과 유지보수성 향상
- **객체지향 프로그래밍(OOP)의 핵심 개념**:
  - **캡슐화(Encapsulation)**: 데이터와 메소드를 하나로 묶어서 관리
  - **상속(Inheritance)**: 기존 클래스를 확장하여 새로운 클래스 생성
  - **다형성(Polymorphism)**: 같은 인터페이스로 다른 동작 수행


### 클래스 정의와 기본 구조
- **클래스 정의**: `class 클래스명:` 형태로 정의
- **클래스 속성**: 클래스 내부의 변수
- **클래스 메소드**: 클래스 내부의 함수
- **생성자**: `__init__` 메소드로 객체 초기화


```python
# 기본 클래스 정의 예시
class Person:
    # 클래스 속성 (모든 인스턴스가 공유)
    species = "Homo sapiens"
    
    # 생성자 (인스턴스 생성 시 자동 호출)
    def __init__(self, name, age):
        self.name = name  # 인스턴스 속성
        self.age = age    # 인스턴스 속성
    
    # 인스턴스 메소드
    def introduce(self):
        return f"안녕하세요, 저는 {self.name}이고 {self.age}살입니다."
    
    def have_birthday(self):
        self.age += 1
        return f"{self.name}의 생일! 이제 {self.age}살이 되었습니다."

# 객체 생성 및 사용
person1 = Person("김철수", 25)
person2 = Person("이영희", 30)

print(person1.introduce())
print(person2.introduce())
print(person1.have_birthday())
print(f"클래스 속성: {Person.species}")
print(f"인스턴스 속성: {person1.name}, {person1.age}")
```

### 상속(Inheritance)
- **상속**: 기존 클래스의 기능을 확장하여 새로운 클래스 생성
- **부모 클래스(Parent Class)**: 상속을 받는 기본 클래스
- **자식 클래스(Child Class)**: 상속을 받는 새로운 클래스
- **super()**: 부모 클래스의 메소드 호출
- **메소드 오버라이딩**: 부모 클래스의 메소드를 자식 클래스에서 재정의


```python
# 상속 예시
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def make_sound(self):
        return "동물이 소리를 냅니다."
    
    def eat(self):
        return f"{self.name}이(가) 먹이를 먹습니다."
    
    def sleep(self):
        return f"{self.name}이(가) 잠을 잡니다."

class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)  # 부모 클래스 생성자 호출
        self.breed = breed
    
    def make_sound(self):  # 메소드 오버라이딩
        return f"{self.name}이(가) 멍멍 짖습니다."
    
    def fetch(self):
        return f"{self.name}이(가) 공을 가져옵니다."

class Cat(Animal):
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color = color
    
    def make_sound(self):  # 메소드 오버라이딩
        return f"{self.name}이(가) 야옹 웁니다."
    
    def climb(self):
        return f"{self.name}이(가) 나무를 올라갑니다."

# 객체 생성 및 사용
dog = Dog("멍멍이", 3, "골든리트리버")
cat = Cat("야옹이", 2, "검은색")

print(dog.make_sound())
print(dog.eat())
print(dog.fetch())

print(cat.make_sound())
print(cat.sleep())
print(cat.climb())

print(f"강아지 품종: {dog.breed}")
print(f"고양이 색깔: {cat.color}")
```

### 가시성(Visibility)
- **Public**: 이름 앞에 밑줄 없음 (예: `value`) - 어디서나 접근 가능
- **Protected**: 이름 앞에 밑줄 1개 (예: `_value`) - 하위 클래스에서 접근 권장, 외부에서 직접 접근은 권장하지 않음
- **Private**: 이름 앞에 밑줄 2개 (예: `__value`) - 클래스 내부에서만 접근, 이름이 맹글링(name mangling)되어 외부/하위 클래스에서 직접 접근 어렵게 만듦


```python
# 가시성 예제 (PDF 내용 기반)
class Base:
    def __init__(self):
        self.public_value = 1      # public
        self._protected_value = 2  # protected
        self.__private_value = 3   # private

class Child(Base):
    def show(self):
        print(self.public_value)      # 가능
        print(self._protected_value)  # 가능
        # print(self.__private_value) # 불가(AttributeError)

class Other:
    def show(self, obj: Base):
        print(obj.public_value)      # 가능
        print(obj._protected_value)  # 가능(하지만 권장되지 않음)
        # print(obj.__private_value) # 불가(AttributeError)

# 테스트
base = Base()
child = Child()
other = Other()

print("=== Child 클래스에서 접근 ===")
child.show()

print("\n=== Other 클래스에서 접근 ===")
other.show(base)

# 비공개 속성에 직접 접근 시도
try:
    print(base.__private_value)  # AttributeError 발생
except AttributeError as e:
    print(f"비공개 속성 접근 오류: {e}")
```

### 추상 클래스(Abstract Class)
- **추상 메서드**: `abc` 모듈의 `ABC`(Abstract Base Class)와 `@abstractmethod` 데코레이터 사용
- **특징**: 추상 메서드는 반드시 하위 클래스에서 구현해야 하며, 직접 인스턴스화할 수 없음
- **목적**: 인터페이스를 정의하고 하위 클래스에서 반드시 구현하도록 강제


```python
# 추상 클래스 예제 (PDF 내용 기반)
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "멍멍"

class Cat(Animal):
    def sound(self):
        return "야옹"

# 테스트
# a = Animal()  # 에러: 추상클래스는 인스턴스화 불가
d = Dog()
c = Cat()

print(f"개 소리: {d.sound()}")
print(f"고양이 소리: {c.sound()}")

# 추상 클래스 인스턴스화 시도
try:
    animal = Animal()
except TypeError as e:
    print(f"추상 클래스 인스턴스화 오류: {e}")
```

### 정적 멤버(Static Members)
- **클래스 변수**: 클래스 전체에서 공유되는 변수 (정적 멤버)
- **정적 메서드**: `@staticmethod` 데코레이터를 사용해 정의
- **접근 방법**: 클래스.변수 형태로 접근
- **특징**: 인스턴스와 무관하게 클래스 레벨에서 동작


```python
# 정적 멤버 예제 (PDF 내용 기반)
class Counter:
    count = 0  # 클래스 변수(정적 멤버)
    
    def __init__(self):
        self.__object_counter = 1  # private
    
    @staticmethod
    def increment_class():
        Counter.count += 1
    
    def increment_object(self):
        self.__object_counter += 1  # private

# 테스트
Counter.increment_class()
print(f"클래스 카운트: {Counter.count}")  # 1

c = Counter()  # 인스턴스 생성
print(f"객체 카운터: {c._Counter__object_counter}")  # private 접근 (맹글링된 이름 사용)

c.increment_object()
print(f"객체 카운터 증가 후: {c._Counter__object_counter}")

# 클래스 변수는 모든 인스턴스가 공유
c2 = Counter()
Counter.increment_class()
print(f"클래스 카운트: {Counter.count}")  # 2
print(f"c1 객체 카운터: {c._Counter__object_counter}")
print(f"c2 객체 카운터: {c2._Counter__object_counter}")
```

### Dunder Methods (특수 메서드)
- **Dunder**: Double Under lines의 약자 (`__XXX__`)
- **자동 호출**: 특정 상황에서 자동으로 호출되는 메서드
- **주요 Dunder Methods**:
  - `__init__(self, ...)`: 클래스 생성 함수, 객체 만들어질 때 자동 실행
  - `__str__(self)`: 클래스 객체를 문자열로 형변환할 때 자동 실행
  - `__eq__(self, other)`: 이 객체와 다른 객체를 같은지 여부
  - `__gt__(self, other)`: 이 객체가 다른 객체보다 큰지 여부
  - `__del__(self)`: 객체가 삭제될 때 자동 실행


```python
# Dunder Methods 예제 (PDF 내용 기반)
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"Student(name={self.name}, age={self.age})"
    
    def __eq__(self, other):
        if isinstance(other, Student):
            return self.name == other.name and self.age == other.age
        return False
    
    def __del__(self):
        print(f"Student {self.name} is being deleted")

# 테스트
student1 = Student("Alice", 20)
student2 = Student("Bob", 22)
student3 = Student("Alice", 20)

print(student1)  # Calls __str__
print(student2)  # Calls __str__
print(student1 == student2)  # Calls __eq__
print(student1 == student3)  # Calls __eq__

# 객체 삭제
del student1  # Calls __del__

# dir() 함수로 사용 가능한 메서드 확인
print("\n사용 가능한 메서드들:")
methods = [method for method in dir(Student) if not method.startswith('_')]
print(methods)
```

### 파이썬 데코레이터(Decorator)
- **데코레이터**: 함수나 클래스의 동작을 수정하거나 확장할 때 사용하는 문법
- **사용법**: 함수 정의 위에 `@데코레이터이름` 형태로 사용
- **동작**: 해당 함수가 호출될 때 데코레이터 함수 내에서 해당 함수가 실행
- **구조**: wrapper 함수를 정의하여 원본 함수를 감싸는 형태


```python
# 데코레이터 예제 (PDF 내용 기반)
def simple_decorator(func):
    def wrapper():
        print("함수 실행 전")
        result = func()
        print("함수 실행 후")
        return result
    return wrapper

@simple_decorator
def say_hello():
    print("Hello, World!")

# 테스트
print("=== 예제1: 기본 데코레이터 ===")
say_hello()

# 함수 재할당 방법
def original_say_hello():
    print("Hello, World!")

def simple_decorator2(func):
    def wrapper():
        print("함수 실행 전")
        result = func()
        print("함수 실행 후")
        return result
    return wrapper

# 1. 함수 재할당
original_hello = original_say_hello  # 원본 백업
say_hello2 = simple_decorator2(original_say_hello)  # 함수 재할당

# 2. 다른 함수명
decorated_say_hello = simple_decorator2(original_say_hello)  # 다른 함수명

print("\n=== 예제2: 함수 재할당 ===")
say_hello2()

print("\n=== 예제3: 다른 함수명 ===")
decorated_say_hello()
```
