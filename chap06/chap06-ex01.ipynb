* 연습문제 6-8. 위의 Person 클래스를 수정하여 Employee 클래스를 포함하여 새로운 Person 객체가 생성될 때마다 ID가 1001부터 순차적으로 자동 할당되도록 하라. 새로운 Employee 객체가 생성되고 getId()를 호출할 때 ID가 중복되지 않아야 한다.


```python
class Person:
    _id_counter = 1000  # 클래스 변수로 ID 카운터 관리
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person._id_counter += 1  # 새로운 객체 생성 시 ID 증가
        self._id = Person._id_counter
    
    def getId(self):
        print(f"ID: {self._id}")
        return self._id
    
    def getAge(self):
        print(f"나이: {self.age}")
        return self.age
    
    def getName(self):
        print(f"이름: {self.name}")
        return self.name

class Employee(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)  # 부모 클래스 생성자 호출
        self.department = department
    
    def getDepartment(self):
        print(f"부서: {self.department}")
        return self.department
    
    def getInfo(self):
        print(f"직원 정보 - ID: {self._id}, 이름: {self.name}, 나이: {self.age}, 부서: {self.department}")

# 테스트
print("=== Person 객체 생성 ===")
person1 = Person("김철수", 25)
person2 = Person("이영희", 30)

print("Person1 정보:")
person1.getId()
person1.getName()
person1.getAge()

print("\nPerson2 정보:")
person2.getId()
person2.getName()
person2.getAge()

print("\n=== Employee 객체 생성 ===")
employee1 = Employee("박민수", 28, "개발팀")
employee2 = Employee("정수진", 32, "마케팅팀")

print("Employee1 정보:")
employee1.getInfo()

print("\nEmployee2 정보:")
employee2.getInfo()

print(f"\n=== ID 중복 확인 ===")
print(f"Person1 ID: {person1._id}")
print(f"Person2 ID: {person2._id}")
print(f"Employee1 ID: {employee1._id}")
print(f"Employee2 ID: {employee2._id}")
print("모든 ID가 순차적으로 할당되었습니다!")
```

* 연습문제 6-9. 파이썬 언어에서 다음에 대해 조사하고 예제코드를 실행하여 보라
  - 1. 파이썬으로 싱글톤 패턴을 구현하는 방법들
  - 2. 파이썬에서 멀티쓰레드 구현하는 기법
  - 3. 멀티쓰레드에서 공유 변수에 대한 Critical Section을 보호하는 기법


```python
# 1. 싱글톤 패턴 구현 방법들

print("=== 1. 싱글톤 패턴 구현 방법들 ===")

# 방법 1: __new__ 메소드 사용
class Singleton1:
    _instance = None
    
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
    
    def __init__(self):
        if not hasattr(self, 'initialized'):
            self.value = "싱글톤 객체"
            self.initialized = True

# 방법 2: 데코레이터 사용
def singleton(cls):
    instances = {}
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@singleton
class Singleton2:
    def __init__(self):
        self.value = "데코레이터 싱글톤"

# 방법 3: 메타클래스 사용
class SingletonMeta(type):
    _instances = {}
    
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Singleton3(metaclass=SingletonMeta):
    def __init__(self):
        self.value = "메타클래스 싱글톤"

# 테스트
print("싱글톤 패턴 테스트:")
s1_1 = Singleton1()
s1_2 = Singleton1()
print(f"Singleton1 - 같은 객체인가? {s1_1 is s1_2}")

s2_1 = Singleton2()
s2_2 = Singleton2()
print(f"Singleton2 - 같은 객체인가? {s2_1 is s2_2}")

s3_1 = Singleton3()
s3_2 = Singleton3()
print(f"Singleton3 - 같은 객체인가? {s3_1 is s3_2}")

print("\n=== 2. 멀티쓰레드 구현 기법 ===")

import threading
import time

# 기본 쓰레드 예제
def worker(name, delay):
    for i in range(3):
        print(f"쓰레드 {name}: 작업 {i+1}")
        time.sleep(delay)

# 쓰레드 생성 및 실행
thread1 = threading.Thread(target=worker, args=("A", 0.5))
thread2 = threading.Thread(target=worker, args=("B", 0.3))

print("멀티쓰레드 실행:")
thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("\n=== 3. Critical Section 보호 기법 ===")

# 공유 변수
shared_counter = 0
lock = threading.Lock()

def increment_counter(name, iterations):
    global shared_counter
    for i in range(iterations):
        # Critical Section 보호
        with lock:
            temp = shared_counter
            time.sleep(0.001)  # 시뮬레이션을 위한 지연
            shared_counter = temp + 1
        print(f"쓰레드 {name}: 카운터 증가")

# Lock을 사용하지 않은 경우 (문제 상황)
def increment_without_lock(name, iterations):
    global shared_counter
    for i in range(iterations):
        temp = shared_counter
        time.sleep(0.001)
        shared_counter = temp + 1

print("Critical Section 보호 테스트:")
shared_counter = 0
threads = []
for i in range(3):
    t = threading.Thread(target=increment_counter, args=(f"Thread-{i+1}", 5))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print(f"Lock 사용 후 최종 카운터 값: {shared_counter}")

# Lock을 사용하지 않은 경우의 문제점 시연
print("\nLock을 사용하지 않은 경우의 문제점:")
shared_counter = 0
threads = []
for i in range(3):
    t = threading.Thread(target=increment_without_lock, args=(f"Thread-{i+1}", 5))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print(f"Lock 미사용 후 최종 카운터 값: {shared_counter}")
print("(예상값: 15, 실제값이 다를 수 있음 - 경쟁 상태 때문)")
```

* 연습문제 6-10. 추상클래스 다루기
  - 추상 클래스 `Animal`을 정의하고 `cry` (울음)을 추상 메서드로 정의
  - `Animal`을 상속받는 구체 클래스 `Dog` (개)와 `Cat` (고양이) 정의
  - 각 동물의 울음소리를 출력하는 프로그램 작성
  - 각 동물 객체는 `name` 속성을 가져야 함
  - 생성자에서 `name` 초기화
  - `Animal` 클래스는 현재 생성된 동물 객체의 수를 추적하는 클래스 변수 포함
  - 현재 동물 객체 수를 반환하는 정적 메서드 구현
  - 객체 생성 시 (`__init__`) 카운트 증가, 객체 삭제 시 (`__del__`) 카운트 감소
  - 리스트에 `Dog`와 `Cat` 객체 생성 후 `Animal` 변수로 받아 동적 바인딩 확인
  - 모든 객체가 울도록 하는 함수 작성 후 테스트


```python
# 연습문제 6-10 해답
from abc import ABC, abstractmethod

class Animal(ABC):
    animal_count = 0                        # 클래스 변수로 동물 객체 수를 지정

    def __init__(self):
        Animal.animal_count += 1            # 객체 생성 시 카운트 증가

    @abstractmethod
    def sound(self):
        pass

    @staticmethod
    def get_animal_count():
        return Animal.animal_count          # 현재 동물 객체 수 반환
    
    def __del__(self):
        Animal.animal_count -= 1            # 객체 삭제 시 카운트 감소

class Dog(Animal):
    def __init__(self, name):               # 생성자에서 이름을 초기화 
        super().__init__()                  # 부모 클래스의 생성자 호출
        self.__name = name

    def sound(self):                        # 추상 메소드 구현
        return f"{self.__name}이 멍멍하고 울어요"

class Cat(Animal):
    def __init__(self, name):
        super().__init__()
        self.__name = name
    
    def sound(self):                        # 추상 메소드 구현
        return f"{self.__name}이 야옹하고 울어요"

# 모든 동물이 울도록 하는 함수
def make_all_animals_cry(animals):
    for animal in animals:
        print(animal.sound())

if __name__ == "__main__":
    print("=== 연습문제 6-10 테스트 ===")
    
    # 동물 객체들 생성
    mydog : Dog = Dog("멍멍이")
    mycat : Cat = Cat("야옹이")
    mydog2 : Dog = Dog("바둑이")
    mycat2 : Cat = Cat("나비")
    
    print(mydog.sound())
    print(mycat.sound())
    print(mydog2.sound())
    print(mycat2.sound())
    
    print("전체 동물 수 : ", Animal.get_animal_count())
    
    print("\n=== 동적 바인딩 테스트 ===")
    # Animal 변수로 받아서 동적 바인딩 확인
    animal_refs = [Dog("동적멍멍"), Cat("동적야옹")]
    for animal_ref in animal_refs:
        print(animal_ref.sound())
    
    print("\n=== 모든 동물이 울기 ===")
    animals = [mydog, mycat, mydog2, mycat2]
    make_all_animals_cry(animals)
    
    print("\n현재 동물 객체 수: ", Animal.get_animal_count())
    
    # 객체 삭제 테스트
    print("\n=== 객체 삭제 테스트 ===")
    del mydog  # 멍멍이 삭제
    print("멍멍이 삭제 후 동물 객체 수: ", Animal.get_animal_count())
    
    del mycat2  # 나비 삭제
    print("나비 삭제 후 동물 객체 수: ", Animal.get_animal_count())
    
    # 추상 클래스 인스턴스화 시도 (에러 발생)
    print("\n=== 추상 클래스 인스턴스화 시도 ===")
    try:
        anymal = Animal()  # 추상 클래스는 객체 생성 불가
    except TypeError as e:
        print("추상 클래스 인스턴스화 오류:", e)
```
