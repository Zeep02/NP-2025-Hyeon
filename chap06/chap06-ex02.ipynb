**연습문제 6-3. Moneybox 클래스**

Moneybox 클래스를 작성하라. 이 클래스는:
- 저축한 총액을 저장하는 인스턴스 변수가 있다
- `deposit(money)` 메서드는 인수로 주어진 금액만큼 저축하고 총액을 출력한다
- `extract(money)` 메소드는 money를 인출하고 총액을 출력한다
- `get_total()` 메소드는 총액을 알려준다
- `my_money = Moneybox()` 객체를 생성하고 1000원을 저축하고, 200원을 인출한 다음, 총액을 출력하라


````python
class Moneybox:
    def __init__(self):
        self.__total = 0  # 저축한 총액 (private)
    
    def deposit(self, money):
        """금액을 저축하고 총액을 출력"""
        self.__total += money
        print(f"{money}원 저축. 총액: {self.__total}원")
    
    def extract(self, money):
        """금액을 인출하고 총액을 출력"""
        if self.__total >= money:
            self.__total -= money
            print(f"{money}원 인출. 총액: {self.__total}원")
        else:
            print(f"잔액 부족! 현재 총액: {self.__total}원")
    
    def get_total(self):
        """총액을 반환"""
        return self.__total

# 테스트
my_money = Moneybox()
my_money.deposit(1000)
my_money.extract(200)
print(f"최종 총액: {my_money.get_total()}원")
````

**연습문제 6-4. Password_manager 클래스**

Password_manager 클래스를 작성하라:
- 사용자의 모든 과거 패스워드를 저장하는 `old_passwords` 리스트 변수를 갖는다
- 리스트의 마지막 요소는 현재 패스워드이다
- `get_password`: 현재 패스워드를 알려주는 메소드
- `set_password`: 새로운 패스워드를 설정하는 메소드 (과거에 사용하지 않은 새로운 패스워드만 유효)
- `is_correct`: 문자열을 입력받아 현재 패스워드와 같은지 True/False 반환


````python
class Password_manager:
    def __init__(self, initial_password):
        self.old_passwords = [initial_password]  # 초기 패스워드
    
    def get_password(self):
        """현재 패스워드 반환"""
        return self.old_passwords[-1]
    
    def set_password(self, new_password):
        """새로운 패스워드 설정"""
        if new_password in self.old_passwords:
            print("과거에 사용한 패스워드입니다. 다른 패스워드를 사용하세요.")
            return False
        else:
            self.old_passwords.append(new_password)
            print("패스워드가 변경되었습니다.")
            return True
    
    def is_correct(self, password):
        """현재 패스워드와 일치하는지 확인"""
        return password == self.get_password()

# 테스트
pm = Password_manager("password123")
print(f"현재 패스워드: {pm.get_password()}")
print(f"패스워드 확인 (password123): {pm.is_correct('password123')}")
print(f"패스워드 확인 (wrong): {pm.is_correct('wrong')}")

pm.set_password("newpass456")
print(f"현재 패스워드: {pm.get_password()}")

pm.set_password("password123")  # 과거 패스워드 재사용 시도
pm.set_password("anotherpass789")  # 새로운 패스워드 설정

print(f"패스워드 히스토리: {pm.old_passwords}")
````

**연습문제 6-5. Articles 클래스**

Articles 클래스를 작성하라:
- 제품의 이름, 재고 수량, 단가를 나타내는 `name`, `stock`, `price` 인스턴스 변수를 갖는다
- `offer_price()` 메소드: 구매 수량을 인수로 전달받아 총 가격을 반환
  - 1~10개 이하: 정상 가격
  - 11~50개: 5% 할인
  - 51개 이상: 10% 할인
- `purchase()` 메소드: 구매 수량을 인수로 전달받아 재고 수량을 감소
- 단가 1000원인 "mask"를 10,000장으로 설정하고, 40장 구매 가격과 구매 후 재고를 출력


````python
class Articles:
    def __init__(self, name, stock, price):
        self.name = name
        self.stock = stock
        self.price = price
    
    def offer_price(self, quantity):
        """구매 수량에 따른 총 가격 계산 (할인 적용)"""
        total = self.price * quantity
        
        if 1 <= quantity <= 10:
            discount = 0  # 할인 없음
        elif 11 <= quantity <= 50:
            discount = 0.05  # 5% 할인
        else:  # 51개 이상
            discount = 0.10  # 10% 할인
        
        discounted_price = total * (1 - discount)
        return discounted_price
    
    def purchase(self, quantity):
        """구매 수량만큼 재고 감소"""
        if self.stock >= quantity:
            self.stock -= quantity
            price = self.offer_price(quantity)
            print(f"{quantity}개 구매. 가격: {price:.0f}원")
            print(f"남은 재고: {self.stock}개")
            return price
        else:
            print(f"재고 부족! 현재 재고: {self.stock}개")
            return 0

# 테스트
mask = Articles("mask", 10000, 1000)
print(f"제품: {mask.name}, 재고: {mask.stock}개, 단가: {mask.price}원")

# 40장 구매 (5% 할인 적용)
print("\n40장 구매:")
mask.purchase(40)
````

**연습문제 6-6. Student 클래스 (학년 판별)**

다음 프로그램을 실행했을 때 아래 내용이 출력되도록 클래스를 정의하라:
- 학번(id) 앞 4자리가 2018보다 작으면 `grade()` 메소드에 의해 "3학년입니다"가 출력
- 아니면 "3학년이 아닙니다"가 출력

```python
man = Student(name="홍길동", id=20150001, dept="로봇공학과")
print("이름: {}, 학번: {}, 학과: {}".format(man.name, man.id, man.dept))
man.grade()
```

출력:
```
이름: 홍길동, 학번: 20150001, 학과: 로봇공학과
3학년입니다
```


````python
class Student:
    def __init__(self, name, id, dept):
        self.name = name
        self.id = id
        self.dept = dept
    
    def grade(self):
        """학번 앞 4자리로 학년 판별"""
        year = self.id // 10000  # 학번 앞 4자리 추출
        if year < 2018:
            print("3학년입니다")
        else:
            print("3학년이 아닙니다")

# 테스트
man = Student(name="홍길동", id=20150001, dept="로봇공학과")
print("이름: {}, 학번: {}, 학과: {}".format(man.name, man.id, man.dept))
man.grade()

# 다른 학생 테스트
student2 = Student(name="김철수", id=20200001, dept="컴퓨터공학과")
print("\n이름: {}, 학번: {}, 학과: {}".format(student2.name, student2.id, student2.dept))
student2.grade()
````

**연습문제 6-7. Employee 클래스**

다음 Person 클래스를 상속받는 Employee 클래스를 정의하라:
- Employee 클래스에 `employeeID` 속성을 추가
- `getID()` 메소드를 정의 (employeeID를 반환)
- `Employee("동양", 65, 2019)`로 생성된 객체의 이름, 나이, ID를 출력

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def getName(self):
        print(self.name)
    
    def getAge(self):
        print(self.age)
```


````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def getName(self):
        print(self.name)
        return self.name
    
    def getAge(self):
        print(self.age)
        return self.age

class Employee(Person):
    def __init__(self, name, age, employeeID):
        super().__init__(name, age)
        self.employeeID = employeeID
    
    def getID(self):
        print(f"직원 ID: {self.employeeID}")
        return self.employeeID

# 테스트
emp = Employee("동양", 65, 2019)
print("이름:", emp.getName())
print("나이:", emp.getAge())
print("ID:", emp.getID())
````
