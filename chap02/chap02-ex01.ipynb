* 연습문제 1. 키보드에서 임의의 두 실수를 입력 받아 곱한 후 그 겨로가를 출력하는 프로그램을 작성하고 실행하라.

```python
x = float(input("첫 번째 수: "))
y = float(input("두 번째 수: "))
z = x * y

print(f"{x} * {y} = {z:.1f}")
```

* 연습문제 2. 1마일은 1.609km이다. 키보드에서 마일(정수)을 입력 받아 km로 변환해 출력하는 프로그램을 작성하라.

```python
x = int(input("마일 입력: "))
y = float(x)*1.609
print(f"{x}마일은 {y:.2f}km입니다.")
```

* 연습문제 3. 원의 반지름을 입력 받아 원의 둘레와 원의 면적을 출력하는 프로그램을 작성하라.

```python
PI = 3.14
x = float(input("원의 반지름 입력: "))
y = 2 * PI * x
z = PI * x ** 2
print(f"원의 둘레: {y:.2f}")
print(f"원의 면적: {z:.2f}")
```
