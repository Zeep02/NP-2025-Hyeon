* 연습문제 1. 초를 입력하면 분과 초로 표시하는 프로그램. (ex. 200초 입력 -> 3분 20초)

```python
x = int(input("초 입력(정수): "))
min = x // 60
sec = x % 60
print(f"{min}분 {sec}초")
```

* 연습문제 2. 분을 입력하면 일, 시간, 분으로 출력하는 프로그램을 만들어라.

```python
x = int(input("분 입력(정수): "))
day = x // 1440
hour = (x % 1440) // 60
min = x % 60

print(f"{day}일 {hour}시 {min}분")
```

* 연습문제 3. 500만원을 년이율 5%로 복리 저금했을 때 5년 후의 원리금의 합계를 출력하는 프로그램

```python
x = 5000000
for i in range(0,5):
    x *= 1.05
print(f"5년후: {x}원")
```

* 연습문제 4. 1부터 n까지의 합은 n(n+1)/2로 주어진다. 1부터 100까지의 합을 구하여 출력하는 프로그램을 작성하고 실행하라.

```python
n = 100
print(f"{n*(n+1)/2}")
```

* 연습문제 5. 판매자가 딸기와 포도를 판매하고 있다. 포도 한 알의 무게는 75g이고 딸기 한 알의 무게는 113.5g이다. 사용자로부터 포도 알의 개수와 딸기의 개수를 입력 받아 총 무게를 계산하여 출력 하는 프로그램을 작성하고 실행하라.

```python
x = int(input("포도 알의 개수 입력: "))
y = int(input("딸기 개수 입력: "))

z = x * 75 + y * 113.5
print(f"총 무게는 {z}g 입니다.")
```
