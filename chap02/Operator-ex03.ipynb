* 연습문제 2. 문자열 "apple"이 세트 {"orange", "banana", "kiwi"}에 포함되어 있는지 검사하는 프로그램을 작성하라.

* 연습문제 3. 숫자 6이 딕셔너리 {"a": 1, "b": 2, "c": 3, "d": 4}의 값들 중 하나인지 검사하는 프로그램을 작성하라.
