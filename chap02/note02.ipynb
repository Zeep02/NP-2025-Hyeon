**서식**|**설명**|**예시**

```python
print(10+20)
print("10"+"20")
print("abc" * 3)
n = 10
print("n=%d" %n)
m = 5.2
print("m=%.1f, n=%d" %(m,n))
```

산술 연산자
| 연산자    | 설명                | 예시         |
|:---------:|:-------------------|:------------|
| x + y     | x와 y를 더함        | c = a + b   |
| x - y     | x에서 y를 뺌        | s = a - b   |
| x * y     | x와 y를 곱함        | m = a * b   |
| x / y     | x를 y로 나눔        | d = a / b   |
| x // y    | x를 y로 나눈 몫     | d = a // b  |
| x % y     | x를 y로 나눈 나머지 | r = a % b   |
| x ** y    | x의 y 제곱          | r = 2 ** 5  |

```python
a = 10
b = 3
c = a % b
d = a // b
e = a / b

print(f"{c}\n{d}\n{e}")
```

관계 연산자
| 연산자   | 설명                        | 사용 예                       |
|:--------:|:--------------------------:|:-----------------------------:|
| x < y    | x가 y보다 작으면 True      | x=1; y=2; print(x<y)          |
| x <= y   | x가 y보다 작거나 같으면 True| x=1; y=2; print(x<=y)         |
| x > y    | x가 y보다 크면 True        | x=1; y=2; print(x>y)          |
| x >= y   | x가 y보다 크거나 같으면 True| x=1; y=2; print(x>=y)         |
| x == y   | x와 y가 같으면 True        | x=1; y=2; print(x==y)         |
| x != y   | x와 y가 다르면 True        | x=1; y=2; print(x!=y)         |

```python
score = 95
print(score >= 90)
print(score < 80)

```
